```python
import re
from collections import Counter

# 2. Clean + split into words (this replaces Tokenizer's internal text processing)
def clean_and_tokenize(raw_text):
    raw_text = raw_text.lower()
    words = re.findall(r"[a-zA-Z']+", raw_text)
    return words

words = clean_and_tokenize(cricket_history)

# 3. Create tokenizer and "fit" it on your words (this replaces fit_on_texts)
class WordTokenizer:
    def __init__(self):
        self.word2idx = {}
        self.idx2word = {}
        self.vocab_size = 0

    def fit(self, words):
        counts = Counter(words)
        vocab = sorted(counts, key=counts.get, reverse=True)
        self.word2idx = {w: i + 1 for i, w in enumerate(vocab)}  # 0 reserved for padding
        self.idx2word = {i: w for w, i in self.word2idx.items()}
        self.vocab_size = len(self.word2idx) + 1

    def encode(self, words):
        return [self.word2idx[w] for w in words if w in self.word2idx]

    def decode(self, idx):
        return self.idx2word.get(idx, "")

tokenizer = WordTokenizer()
tokenizer.fit(words)          # <- equivalent of tokenizer.fit_on_texts([cricket_history])

# 4. Convert words to numbers (this replaces texts_to_sequences)
encoded = tokenizer.encode(words)   # <- equivalent of tokenizer.texts_to_sequences([cricket_history])
```

```python
WINDOW = 4
encoded_all = tokenizer.encode(words)
seqs = [encoded_all[i-WINDOW:i+1] for i in range(WINDOW, len(encoded_all))]
seqs = torch.tensor(seqs, dtype=torch.long)

x = seqs[:, :-1]
y = seqs[:, -1]

print(x.shape, y.shape)
```

```python
from torch.utils.data import TensorDataset, DataLoader

loader = DataLoader(TensorDataset(x, y), batch_size=32, shuffle=True)
```

```python
class TextModel(nn.Module):
    def __init__(self, num_classes, embed_dim=48, hidden_size=64):
        super().__init__()
        self.embedding = nn.Embedding(num_classes, embed_dim)
        self.lstm = nn.LSTM(embed_dim, hidden_size, batch_first=True)
        self.dense = nn.Linear(hidden_size, num_classes)

    def forward(self, x):
        x = self.embedding(x)
        x, (hidden, cell) = self.lstm(x)
        return self.dense(hidden[-1])

model = TextModel(num_classes)
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
```

```python
criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)
```

```python
epochs = 100
for epoch in range(epochs):
    model.train()
    total_loss, correct, total = 0, 0, 0

    for batch_x, batch_y in loader:
        outputs = model(batch_x)
        loss = criterion(outputs, batch_y)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item() * batch_x.size(0)
        correct += (torch.argmax(outputs, dim=1) == batch_y).sum().item()
        total += batch_y.size(0)

    if (epoch + 1) % 5 == 0 or epoch == 0:
        print(f"Epoch [{epoch+1}/{epochs}] Loss: {total_loss/total:.4f} Accuracy: {correct/total:.4f}")
```

```python
def predict_next_word(model, tokenizer, seed_text, window):
    model.eval()
    tokens = clean_and_tokenize(seed_text)
    encoded = tokenizer.encode(tokens)
    encoded = ([0]*(window-len(encoded)) + encoded) if len(encoded) < window else encoded[-window:]
    input_tensor = torch.tensor([encoded], dtype=torch.long)
    with torch.no_grad():
        output = model(input_tensor)
        return tokenizer.decode(torch.argmax(output, dim=1).item())


def generate_text(model, tokenizer, seed_text, window, num_words=5):
    text = seed_text
    for _ in range(num_words):
        next_word = predict_next_word(model, tokenizer, text, window)
        if next_word == "":
            break
        text += " " + next_word
    return text


print(generate_text(model, tokenizer, "Twenty Twenty Cricket was ", WINDOW, num_words=5))
```
